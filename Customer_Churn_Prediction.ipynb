```python
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import warnings
warnings.filterwarnings('ignore')
```

**1. OVERALL CHURN STATUS**

```python
data['Churn'].value_counts()
```

```python
labels = data['Churn'].unique()
values = data['Churn'].value_counts()

fig = go.Figure(data=[go.Pie(labels=labels, values=values, hole=.3, textfont_size= 20, title='<b>Churn</b>', titlefont_size=15)])
fig.update_layout(title_text='<b>Churn Distribution</b>', title_font_size=28, title_font_family = 'Georgia')
fig.show()
```

**2. CHURN DISTRIBUTION BY GENDER**

```python
data.Churn[data.Churn=='Yes'].groupby(by = data.gender).count()
```

```python
data.Churn[data.Churn == 'No'].groupby(by = data.gender).count()
```

```python
fig = px.histogram(data, x="Churn", color = "gender", barmode = "group", title = "<b>Churn Distribution By Gender<b>")
fig.update_layout(width=700, height=500, bargap=0.2,  title_font_size=28, title_font_family="Georgia")
fig.show()
```
